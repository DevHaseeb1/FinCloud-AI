### <b>Dataset – Overview</b>

This project utilizes a **real-world AWS Cost and Usage Report (CUR) dataset** generated from my personal AWS account while working on various cloud-based projects and experiments. The billing data was collected directly from AWS Cost Explorer and Cost & Usage Reports, which provide detailed records of resource consumption and associated costs across AWS services.

To prepare the dataset for analysis, the original CUR data was preprocessed and cleaned by selecting the most relevant attributes related to cloud infrastructure, service usage, pricing, resource configurations, and organizational metadata. The resulting billing dataset contains **2,516 records (rows)** and **30 features (columns)**, making it suitable for cloud cost analysis and machine learning applications.

The dataset includes information such as AWS account details, services, regions, resource identifiers, instance specifications, usage quantities, billing costs, pricing models, operating systems, storage configurations, and resource tags. These attributes provide valuable insights into cloud resource consumption patterns and spending behavior.

The primary goal of this dataset is to support **AWS Cloud Cost Optimization** and **FinOps-driven decision making**. 
By integrating billing records with operational usage metrics, the dataset enables:

* **Cost Allocation and Spending Analysis**
* **Resource Utilization Monitoring**
* **Cloud Waste Detection**
* **Idle Resource Identification**
* **Infrastructure Rightsizing Recommendations**
* **Cost Anomaly Detection**
* **Cost Forecasting and Budget Planning**
* **Machine Learning-Based Cost Optimization**
* **Cloud Resource Efficiency Analysis**

#### ***Note*** 

`AWS Cost and Usage Report (CUR)` is a unified billing dataset that combines records from many AWS services into a single table.

Since different AWS services provide different attributes, AWS leaves non-applicable fields **empty (NaN)** instead of filling them with incorrect values.

Therefore, many missing values in CUR are expected behavior and do not indicate data corruption or poor data quality.

* #### **Example 1: Amazon S3**

Amazon S3 is an object storage service.

S3 does not have:

CPU,
Memory,
Operating System,
Instance Type,
vCPU,
Processor,
RAM

Therefore AWS cannot provide values for those columns.

Example:

**service_code	operating_system	memory	instance_type**

   AmazonS3	            NaN	          NaN	     NaN

This is correct because an S3 bucket is not a virtual machine.

* #### **Example 2: CloudWatch**

CloudWatch is a monitoring service.

CloudWatch does not run on:

EC2 instances,
CPUs,
RAM,
Storage devices

Therefore columns such as:

operating_system
memory
vcpu
storage
instance_type

remain NaN.

Example:

**service_code	memory	vcpu	operating_system**

AmazonCloudWatch	NaN	NaN	NaN

This is expected.

* #### **Example 3: AWS Lambda**

Lambda is a serverless service.

AWS manages the infrastructure.

Users do not see:

OS,
Processor,
Storage Device,
Instance Family

Therefore many infrastructure-related columns become NaN.

Example:

**service_code	operating_system	instance_type**

AWSLambda	NaN	NaN

* #### **Example 4: EBS**

Amazon EBS is block storage.

EBS has:

volume_type,
storage

But it does not have:

CPU,
RAM,
Operating System,
Processor

Therefore:

**service_code	memory	operating_system**
AmazonEBS	NaN	NaN

is normal.



* #### **Example 5: storage_class**

Only some storage services support storage classes.

Examples:

S3

Supports:

Standard
Standard-IA
Glacier
Deep Archive

Therefore storage_class may have values.

EC2

EC2 instances do not have S3 storage classes.

Therefore:

storage_class = NaN

is expected.

* #### **Example 6: operation Syatem**

Some CUR records represent:

Reserved Instance charges
Savings Plans charges
Tax records
Credits
Support fees

These billing entries are not actual operations.

Therefore:

operation = NaN

can occur.

* #### **Example 7: resource_id**

Not every AWS charge belongs to a specific resource.

Examples:

Data Transfer
CloudWatch Metrics
Tax
Support
Regional Charges

For these records AWS cannot assign:

resource_id

Therefore:

resource_id = NaN

is completely normal.
